```python
1. Line Plot
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

plt.plot(x, y)

plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)

plt.savefig("line_plot.png")
plt.show()
```

```python
2. Bar Plot
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'Category': ['A', 'B', 'C'],
    'Values': [10, 20, 15]
})

df.plot(kind='bar', x='Category', y='Values')

plt.title("Bar Plot")
plt.xlabel("Category")
plt.ylabel("Values")

plt.show()
```

```python
3. Histogram and Density Plot
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

data = np.random.randn(1000)

sns.histplot(data, kde=True)

plt.title("Histogram and Density")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```

```python
4. Scatter Plot
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x, y)

plt.title("Scatter Plot")
plt.xlabel("x")
plt.ylabel("y")

plt.show()
```

```python
5. Box Plot
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

data = np.random.randn(100, 4)

sns.boxplot(data=data)

plt.title("Box Plot")

plt.show()
```
