```python
import pandas as pd

df = pd.DataFrame({'Marks':[10,20,30,40,100]})

Q1 = df['Marks'].quantile(0.25)
Q3 = df['Marks'].quantile(0.75)
IQR = Q3 - Q1

outliers = df[(df['Marks'] < Q1-1.5*IQR) | (df['Marks'] > Q3+1.5*IQR)]

print("Outliers:", outliers['Marks'].tolist())
print("Sum:", outliers['Marks'].sum())
```
