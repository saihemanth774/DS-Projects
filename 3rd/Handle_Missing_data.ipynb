```python
import pandas as pd
import numpy as np

data = {
    'Name': ['Aravind', 'Ravi', 'Kumar', 'Sita', 'Anu'],
    'Age': [20, np.nan, 21, 22, np.nan],
    'Marks': [85, 90, np.nan, 75, 88]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

print("\nUsing fillna():")
df_fill = df.fillna(0)
print(df_fill)

```
