```python
import pandas as pd

data = {
    'Name': ['Aravind', 'Ravi', 'Kumar', 'Sita'],
    'Age': [20, 21, 22, 23],
    'Marks': [85, 90, 75, 95]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

df['Marks'] = df['Marks'] + 5
df['Result'] = df['Marks'].apply(lambda x: 'Pass' if x >= 40 else 'Fail')

print("\nTransformed Data:")
print(df)
```
